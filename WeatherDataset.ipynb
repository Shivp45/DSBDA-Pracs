```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/vega/vega-datasets/master/data/seattle-weather.csv"
df = pd.read_csv(url)

df.head()
```

```python
print(df.columns)
```

```python
# Average temperature
df['Temperature'] = (df['temp_max'] + df['temp_min']) / 2

# Approximate dew point (for learning purpose)
df['DewPoint'] = df['temp_min'] - 2  

# Wind speed already exists
df['WindSpeed'] = df['wind']
```

```python
avg_temp = df['Temperature'].mean()
avg_dew = df['DewPoint'].mean()
avg_wind = df['WindSpeed'].mean()
```

```python
print(" Weather Averages:\n")

print(f" Temperature: {avg_temp:.2f}")
print(f" Dew Point: {avg_dew:.2f}")
print(f" Wind Speed: {avg_wind:.2f}")
```

```python
plt.figure()

plt.plot(df['Temperature'])
plt.title("Temperature Trend")

plt.xlabel("Days")
plt.ylabel("Temperature")

plt.show()
```
