```python
# Importing required libraries
import pandas as pd
import numpy as np
```

```python
# Load dataset from URL
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

# Display first 5 rows
df.head()
```

```python
# Check dimensions (rows, columns)
print("Shape of dataset:", df.shape)

# Column names
print("Columns:", df.columns)

# Data types
df.info()
```

```python
# Check missing values
print("Missing values:\n", df.isnull().sum())

# Basic statistics
df.describe()
```

```python
# Display all column names with explanation
for col in df.columns:
    print(col)
```

```python
# Check data types
df.dtypes
```

```python
# Convert 'Survived' to categorical
df['Survived'] = df['Survived'].astype('category')

# Convert 'Pclass' to category
df['Pclass'] = df['Pclass'].astype('category')

df.dtypes
```

```python
# Handle missing values properly (without warnings)

# Fill missing Age with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Fill missing Embarked with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Check missing values again
print("Missing values after handling:\n")
print(df.isnull().sum())
```

```python
# Normalize Age and Fare using Min-Max scaling
df['Age_norm'] = (df['Age'] - df['Age'].min()) / (df['Age'].max() - df['Age'].min())

df['Fare_norm'] = (df['Fare'] - df['Fare'].min()) / (df['Fare'].max() - df['Fare'].min())

df.head()
```

```python
# Convert categorical variables using one-hot encoding
df_encoded = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)

df_encoded.head()
```

```python
# Final dataset info
df_encoded.info()

# Final shape
print("Final Shape:", df_encoded.shape)
```
