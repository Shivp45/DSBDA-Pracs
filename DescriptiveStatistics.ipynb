```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

```python
# Create age groups
bins = [0, 12, 20, 40, 60, 100]
labels = ['Child', 'Teen', 'Adult', 'Middle Age', 'Senior']

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

df[['Age', 'Age_Group']].head()
```

```python
# Fill missing Age with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Fill missing Fare with median
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
```

```python
grouped_stats = df.groupby('Age_Group', observed=True)['Fare'].agg(['mean', 'median', 'min', 'max', 'std'])

grouped_stats
```

```python
fare_lists = df.groupby('Age_Group', observed=True)['Fare'].apply(list)

fare_lists
```

```python
iris_url = "https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv"
iris = pd.read_csv(iris_url)

iris.head()
```

```python
iris['species'].unique()
```

```python
iris.groupby('species').describe()
```

```python
species_groups = iris.groupby('species')

for species, data in species_groups:
    print(f"\nSpecies: {species}")
    
    print("Mean:\n", data.mean(numeric_only=True))
    
    print("\nStandard Deviation:\n", data.std(numeric_only=True))
    
    print("\nPercentiles:\n", data.quantile([0.25, 0.5, 0.75], numeric_only=True))
```

```python
setosa = iris[iris['species'] == 'setosa']
versicolor = iris[iris['species'] == 'versicolor']
virginica = iris[iris['species'] == 'virginica']

print("Setosa Stats:\n", setosa.describe())
print("\nVersicolor Stats:\n", versicolor.describe())
print("\nVirginica Stats:\n", virginica.describe())
```
