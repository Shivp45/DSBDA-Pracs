```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')  # <-- important
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
text = """
Text analytics is the process of analyzing textual data to extract useful information.
It includes tokenization, stemming, and lemmatization techniques.
"""
```

```python
tokens = word_tokenize(text)
print("Tokens:", tokens)
```

```python
pos_tags = pos_tag(tokens)
print("POS Tags:", pos_tags)
```

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print("After Stopwords Removal:", filtered_tokens)
```

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

print("Stemmed Words:", stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print("Lemmatized Words:", lemmatized_words)
```

```python
from collections import Counter

tf = Counter(filtered_tokens)
print("Term Frequency:", tf)
```

```python
documents = [
    "Text analytics is the process of analyzing textual data",
    "It includes tokenization and preprocessing techniques"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

print("Feature Names:", vectorizer.get_feature_names_out())
print("TF-IDF Matrix:\n", tfidf_matrix.toarray())
```
